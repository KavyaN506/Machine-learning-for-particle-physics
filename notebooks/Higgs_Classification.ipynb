```python

# ============================================================
# HIGGS DATASET: DATA LOADING AND PREPROCESSING
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Reproducibility
RANDOM_STATE = 42

# Download and load HIGGS dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00280/HIGGS.csv.gz"

column_names = (
    ["label"] +
    [f"feature_{i}" for i in range(1, 29)]
)

print("Loading HIGGS dataset...")

df = pd.read_csv(
    url,
    compression="gzip",
    header=None,
    names=column_names,
    nrows=100000
)

print("Dataset loaded successfully!")

# Separate features and target
X = df.drop("label", axis=1)
y = df["label"]

# Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

# Feature scaling for Logistic Regression
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Dataset information
print("\n========== DATASET INFORMATION ==========")
print("Dataset shape:", df.shape)
print("Features:", X.shape)
print("Target:", y.shape)

print("\nTraining features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining labels:", y_train.shape)
print("Testing labels:", y_test.shape)

print("\nClass distribution:")
print(y.value_counts())

print("\nData preparation completed successfully!")
```

Day 6 — Logistic Regression and Decision Tree

```python

# ============================================================
# MACHINE LEARNING FOR PARTICLE PHYSICS
# MODEL TRAINING, EVALUATION AND HYPERPARAMETER TUNING
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import (
    StratifiedKFold,
    cross_validate,
    RandomizedSearchCV
)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    ConfusionMatrixDisplay
)

# ============================================================
# 1. VERIFY DATA
# ============================================================

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)

print("Scaled training features:", X_train_scaled.shape)
print("Scaled testing features:", X_test_scaled.shape)


# ============================================================
# 2. LOGISTIC REGRESSION
# ============================================================

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train_scaled, y_train)

lr_predictions = lr_model.predict(X_test_scaled)
lr_probabilities = lr_model.predict_proba(X_test_scaled)[:, 1]

print("\nLogistic Regression training completed!")


# ============================================================
# 3. DECISION TREE
# ============================================================

dt_model = DecisionTreeClassifier(
    max_depth=10,
    min_samples_leaf=5,
    random_state=42
)

dt_model.fit(X_train, y_train)

dt_predictions = dt_model.predict(X_test)
dt_probabilities = dt_model.predict_proba(X_test)[:, 1]

print("Decision Tree training completed!")


# ============================================================
# 4. RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=5,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest training completed!")


# ============================================================
# 5. IMPROVED MODEL EVALUATION
# ============================================================

models = {
    "Logistic Regression": (
        lr_model, X_train_scaled, X_test_scaled
    ),
    "Decision Tree": (
        dt_model, X_train, X_test
    ),
    "Random Forest": (
        rf_model, X_train, X_test
    )
}

evaluation_results = []

for name, (model, X_tr, X_te) in models.items():

    train_pred = model.predict(X_tr)
    test_pred = model.predict(X_te)
    test_prob = model.predict_proba(X_te)[:, 1]

    evaluation_results.append({
        "Model": name,
        "Train Accuracy": accuracy_score(y_train, train_pred),
        "Test Accuracy": accuracy_score(y_test, test_pred),
        "Test Precision": precision_score(
            y_test, test_pred, zero_division=0
        ),
        "Test Recall": recall_score(
            y_test, test_pred, zero_division=0
        ),
        "Test F1": f1_score(
            y_test, test_pred, zero_division=0
        ),
        "Test ROC-AUC": roc_auc_score(y_test, test_prob)
    })

evaluation_df = pd.DataFrame(evaluation_results)

print("\n========== TRAINING VS TESTING PERFORMANCE ==========")
print(evaluation_df.round(4).to_string(index=False))


# ============================================================
# 6. STRATIFIED 5-FOLD CROSS-VALIDATION
# ============================================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Pipeline prevents data leakage during scaling
cv_models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]),
    "Decision Tree": DecisionTreeClassifier(
        max_depth=10,
        min_samples_leaf=5,
        random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=15,
        min_samples_leaf=5,
        max_features="sqrt",
        random_state=42,
        n_jobs=1
    )
}

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

cv_results = []

for name, model in cv_models.items():

    print(f"\nEvaluating {name} using 5-fold CV...")

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Mean Accuracy": scores["test_accuracy"].mean(),
        "Accuracy Std": scores["test_accuracy"].std(),
        "Mean Precision": scores["test_precision"].mean(),
        "Mean Recall": scores["test_recall"].mean(),
        "Mean F1": scores["test_f1"].mean(),
        "Mean ROC-AUC": scores["test_roc_auc"].mean(),
        "ROC-AUC Std": scores["test_roc_auc"].std()
    })

cv_df = pd.DataFrame(cv_results)

print("\n========== STRATIFIED 5-FOLD CROSS-VALIDATION ==========")
print(cv_df.round(4).to_string(index=False))


# ============================================================
# 7. RANDOM FOREST HYPERPARAMETER TUNING
# ============================================================

param_grid = {
    "n_estimators": [100, 150],
    "max_depth": [10, 15, 20],
    "min_samples_leaf": [5, 10],
    "max_features": ["sqrt"]
}

rf_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=1
    ),
    param_distributions=param_grid,
    n_iter=4,
    scoring="roc_auc",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

print("\nStarting Random Forest hyperparameter tuning...")

rf_search.fit(X_train, y_train)

print("\n========== HYPERPARAMETER TUNING RESULTS ==========")

print("Best parameters:")
print(rf_search.best_params_)

print("\nBest cross-validation ROC-AUC:")
print(round(rf_search.best_score_, 4))


# ============================================================
# 8. EVALUATE TUNED RANDOM FOREST
# ============================================================

best_rf_model = rf_search.best_estimator_

best_rf_predictions = best_rf_model.predict(X_test)

best_rf_probabilities = (
    best_rf_model.predict_proba(X_test)[:, 1]
)

print("\n========== TUNED RANDOM FOREST TEST RESULTS ==========")

print(
    "Accuracy:",
    round(accuracy_score(y_test, best_rf_predictions), 4)
)

print(
    "Precision:",
    round(precision_score(
        y_test, best_rf_predictions, zero_division=0
    ), 4)
)

print(
    "Recall:",
    round(recall_score(
        y_test, best_rf_predictions, zero_division=0
    ), 4)
)

print(
    "F1-score:",
    round(f1_score(
        y_test, best_rf_predictions, zero_division=0
    ), 4)
)

print(
    "ROC-AUC:",
    round(roc_auc_score(y_test, best_rf_probabilities), 4)
)

print("\nClassification Report:")
print(classification_report(
    y_test,
    best_rf_predictions,
    zero_division=0
))


# ============================================================
# 9. COMPARE BASELINE AND TUNED RANDOM FOREST
# ============================================================

rf_comparison = pd.DataFrame({
    "Model": [
        "Baseline Random Forest",
        "Tuned Random Forest"
    ],
    "Accuracy": [
        accuracy_score(y_test, rf_predictions),
        accuracy_score(y_test, best_rf_predictions)
    ],
    "Precision": [
        precision_score(y_test, rf_predictions, zero_division=0),
        precision_score(y_test, best_rf_predictions, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, rf_predictions, zero_division=0),
        recall_score(y_test, best_rf_predictions, zero_division=0)
    ],
    "F1-score": [
        f1_score(y_test, rf_predictions, zero_division=0),
        f1_score(y_test, best_rf_predictions, zero_division=0)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, rf_probabilities),
        roc_auc_score(y_test, best_rf_probabilities)
    ]
})

print("\n========== BASELINE VS TUNED RANDOM FOREST ==========")
print(rf_comparison.round(4).to_string(index=False))


# ============================================================
# 10. CLASSIFICATION REPORTS
# ============================================================

print("\nLOGISTIC REGRESSION REPORT")
print(classification_report(y_test, lr_predictions))

print("\nDECISION TREE REPORT")
print(classification_report(y_test, dt_predictions))

print("\nRANDOM FOREST REPORT")
print(classification_report(y_test, rf_predictions))


# ============================================================
# 11. CONFUSION MATRICES
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

ConfusionMatrixDisplay.from_predictions(
    y_test, lr_predictions,
    ax=axes[0, 0], cmap="Blues"
)
axes[0, 0].set_title("Logistic Regression")

ConfusionMatrixDisplay.from_predictions(
    y_test, dt_predictions,
    ax=axes[0, 1], cmap="Greens"
)
axes[0, 1].set_title("Decision Tree")

ConfusionMatrixDisplay.from_predictions(
    y_test, rf_predictions,
    ax=axes[1, 0], cmap="Oranges"
)
axes[1, 0].set_title("Baseline Random Forest")

ConfusionMatrixDisplay.from_predictions(
    y_test, best_rf_predictions,
    ax=axes[1, 1], cmap="Purples"
)
axes[1, 1].set_title("Tuned Random Forest")

plt.tight_layout()
plt.show()


# ============================================================
# 12. FINAL MODEL COMPARISON
# ============================================================

comparison = evaluation_df[
    ["Model", "Test Accuracy", "Test F1", "Test ROC-AUC"]
].sort_values(
    by="Test ROC-AUC",
    ascending=False
)

print("\n========== FINAL MODEL COMPARISON ==========")
print(comparison.round(4).to_string(index=False))

print("\nModel training and evaluation completed successfully!")
```
