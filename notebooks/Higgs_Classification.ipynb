Day 3 - NumPy and Pandas

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create dataset
data = {
    "event_id": [1, 2, 3, 4, 5, 6],
    "energy": [90, 110, 150, 170, 130, 200],
    "momentum": [15, 20, 25, 30, 22, 35],
    "label": [0, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)

# 1. Display first five rows
print(df.head())

# 2. Calculate average energy
def calculate_average(values):
    return np.mean(values)

print("Average energy:", calculate_average(df["energy"]))

# 3. Maximum energy
print("Maximum energy:", np.max(df["energy"]))

# 4. Filter energy greater than 130
print(df[df["energy"] > 130])

# 5. Count signal events
signal_count = (df["label"] == 1).sum()
print("Signal events:", signal_count)

# 6. Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["energy"], bins=5)
plt.xlabel("Particle Energy")
plt.ylabel("Number of Events")
plt.title("Distribution of Particle Energy")
plt.show()

# 7. Calculate average momentum
print("Average momentum:", calculate_average(df["momentum"]))
```

Day 5 - Data Preprocessing

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/higgs_sample_100k.csv")

print(df.shape)
print(df.head())

from google.colab import files

uploaded = files.upload()

df = pd.read_csv("higgs_sample_100k.csv")

# Check dimensions
print("Shape:", df.shape)

# Check missing values
print("Missing values:", df.isnull().sum().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check data types
print(df.dtypes.value_counts())

print(df["label"].value_counts())
print(df["label"].value_counts(normalize=True) * 100)

# Separate input features
X = df.drop("label", axis=1)

# Separate target variable
y = df["label"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print(X.head())
print(y.head())

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit scaler using training data only
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Scaled training data:", X_train_scaled.shape)
print("Scaled testing data:", X_test_scaled.shape)

print("Training mean:", np.mean(X_train_scaled))
print("Training standard deviation:", np.std(X_train_scaled))

print("Testing data shape:", X_test_scaled.shape)

import joblib

joblib.dump(scaler, "/content/higgs_scaler.pkl")

print("Scaler saved successfully!")
```

Day 6 — Logistic Regression and Decision Tree

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)

print("Scaled training features:", X_train_scaled.shape)
print("Scaled testing features:", X_test_scaled.shape)

# Create Logistic Regression model

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model

lr_model.fit(X_train_scaled, y_train)

print("Logistic Regression training completed!")

# Predict on test data

lr_predictions = lr_model.predict(X_test_scaled)

print(lr_predictions[:20])

# Create Decision Tree model

dt_model = DecisionTreeClassifier(
    max_depth=10,
    min_samples_leaf=5,
    random_state=42
)

# Train the model

dt_model.fit(X_train, y_train)

print("Decision Tree training completed!")

# Predict on test data

dt_predictions = dt_model.predict(X_test)

print(dt_predictions[:20])

# Calculate accuracy

lr_accuracy = accuracy_score(y_test, lr_predictions)
dt_accuracy = accuracy_score(y_test, dt_predictions)

print("Logistic Regression Accuracy:", lr_accuracy)
print("Decision Tree Accuracy:", dt_accuracy)

print("LOGISTIC REGRESSION REPORT")
print(classification_report(y_test, lr_predictions))

print("DECISION TREE REPORT")
print(classification_report(y_test, dt_predictions))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Logistic Regression

ConfusionMatrixDisplay.from_predictions(
    y_test,
    lr_predictions,
    ax=axes[0],
    cmap="Blues"
)

axes[0].set_title("Logistic Regression")

# Decision Tree

ConfusionMatrixDisplay.from_predictions(
    y_test,
    dt_predictions,
    ax=axes[1],
    cmap="Greens"
)

axes[1].set_title("Decision Tree")

plt.tight_layout()
plt.show()

comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [lr_accuracy, dt_accuracy]
})

print(comparison)
```

Day 7 — Random Forest Classification! 🌳


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
import pandas as pd

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=5,
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

# Train Random Forest

rf_model.fit(X_train, y_train)

print("Random Forest training completed!")

# Predict on test data

rf_predictions = rf_model.predict(X_test)

print("First 20 predictions:")
print(rf_predictions[:20])

rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Accuracy:", rf_accuracy)

print("RANDOM FOREST CLASSIFICATION REPORT")

print(classification_report(
    y_test,
    rf_predictions
))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    rf_predictions,
    cmap="Purples"
)

plt.title("Random Forest Confusion Matrix")
plt.show()

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        rf_accuracy
    ]
})

comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
)

print(comparison)

plt.figure(figsize=(9, 5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Accuracy")
plt.title("Model Accuracy Comparison")

plt.ylim(0, 1)
plt.xticks(rotation=15)

plt.show()

day7_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        rf_accuracy
    ]
})

day7_results.to_csv(
    "/content/day7_model_comparison.csv",
    index=False
)

print(day7_results)
```

Day 8 — Advanced Model Evaluation

```python
from sklearn.metrics import (
    roc_curve,
    roc_auc_score
)

import matplotlib.pyplot as plt
import pandas as pd

# Logistic Regression probabilities
lr_probabilities = lr_model.predict_proba(X_test_scaled)[:, 1]

# Decision Tree probabilities
dt_probabilities = dt_model.predict_proba(X_test)[:, 1]

# Random Forest probabilities
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

print("Logistic Regression:", lr_probabilities[:10])
print("Decision Tree:", dt_probabilities[:10])
print("Random Forest:", rf_probabilities[:10])

# Calculate ROC-AUC

lr_auc = roc_auc_score(y_test, lr_probabilities)

dt_auc = roc_auc_score(y_test, dt_probabilities)

rf_auc = roc_auc_score(y_test, rf_probabilities)

print("Logistic Regression AUC:", lr_auc)
print("Decision Tree AUC:", dt_auc)
print("Random Forest AUC:", rf_auc)

# Logistic Regression
lr_fpr, lr_tpr, _ = roc_curve(y_test, lr_probabilities)

# Decision Tree
dt_fpr, dt_tpr, _ = roc_curve(y_test, dt_probabilities)

# Random Forest
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probabilities)

plt.figure(figsize=(9, 7))

plt.plot(
    lr_fpr,
    lr_tpr,
    label=f"Logistic Regression (AUC = {lr_auc:.3f})"
)

plt.plot(
    dt_fpr,
    dt_tpr,
    label=f"Decision Tree (AUC = {dt_auc:.3f})"
)

plt.plot(
    rf_fpr,
    rf_tpr,
    label=f"Random Forest (AUC = {rf_auc:.3f})"
)

# Random classification baseline
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier (AUC = 0.500)"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve: Higgs Signal vs Background Classification")

plt.legend()
plt.grid(alpha=0.3)

plt.show()

today_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        rf_accuracy
    ],

    "ROC-AUC": [
        lr_auc,
        dt_auc,
        rf_auc
    ]
})

today_comparison = today_comparison.sort_values(
    by="ROC-AUC",
    ascending=False
)

print(today_comparison)

plt.figure(figsize=(9, 5))

plt.bar(
    today_comparison["Model"],
    today_comparison["ROC-AUC"]
)

plt.xlabel("Machine Learning Models")
plt.ylabel("ROC-AUC Score")
plt.title("ROC-AUC Comparison")

plt.ylim(0, 1)
plt.xticks(rotation=15)

plt.show()

today_comparison.to_csv(
    "/content/today_model_evaluation.csv",
    index=False
)

print("Results saved successfully!")

plt.figure(figsize=(9, 7))

plt.plot(lr_fpr, lr_tpr, label=f"Logistic Regression (AUC={lr_auc:.3f})")
plt.plot(dt_fpr, dt_tpr, label=f"Decision Tree (AUC={dt_auc:.3f})")
plt.plot(rf_fpr, rf_tpr, label=f"Random Forest (AUC={rf_auc:.3f})")
plt.plot([0, 1], [0, 1], "--", label="Random baseline")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - Higgs Classification")
plt.legend()
plt.grid(alpha=0.3)

plt.savefig(
    "/content/today_roc_curves.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Day 9 — Feature Importance, Model Interpretation, and Scientific Limitations. 🔬


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Extract feature importance
rf_importance = rf_model.feature_importances_

# Create a DataFrame
feature_importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": rf_importance
})

# Sort from highest to lowest
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance_df)

top_15_features = feature_importance_df.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_15_features["Feature"],
    top_15_features["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Particle Physics Features")
plt.title("Top 15 Important Features - Random Forest")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

lr_coefficients = lr_model.coef_[0]

lr_importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Coefficient": lr_coefficients,
    "Absolute Importance": np.abs(lr_coefficients)
})

lr_importance_df = lr_importance_df.sort_values(
    by="Absolute Importance",
    ascending=False
)

print(lr_importance_df)

top_15_lr = lr_importance_df.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_15_lr["Feature"],
    top_15_lr["Coefficient"]
)

plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Features")
plt.title("Top 15 Logistic Regression Coefficients")

plt.axvline(0, linestyle="--")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()

from sklearn.inspection import permutation_importance

# Select a subset of test data
X_test_sample = X_test.iloc[:5000]
y_test_sample = y_test.iloc[:5000]

# Calculate permutation importance
permutation_result = permutation_importance(
    rf_model,
    X_test_sample,
    y_test_sample,
    scoring="roc_auc",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

# Create results DataFrame
permutation_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": permutation_result.importances_mean,
    "Standard Deviation": permutation_result.importances_std
})

permutation_df = permutation_df.sort_values(
    by="Importance",
    ascending=False
)

print(permutation_df.head(15))

top_15_permutation = permutation_df.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_15_permutation["Feature"],
    top_15_permutation["Importance"],
    xerr=top_15_permutation["Standard Deviation"]
)

plt.xlabel("Decrease in ROC-AUC after permutation")
plt.ylabel("Features")
plt.title("Permutation Feature Importance - Random Forest")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Save Random Forest importance
feature_importance_df.to_csv(
    "/content/random_forest_feature_importance.csv",
    index=False
)

# Save Logistic Regression coefficients
lr_importance_df.to_csv(
    "/content/logistic_regression_coefficients.csv",
    index=False
)

# Save permutation importance
permutation_df.to_csv(
    "/content/permutation_feature_importance.csv",
    index=False
)

print("Results saved successfully!")
```
